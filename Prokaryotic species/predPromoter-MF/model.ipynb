```python
def getoptimPromoter(fea, sigma):
    """Select optimal features for each sigma type with consistent scaling"""
    model_dir = "./model"
    
    # Apply StandardScaler to ALL feature sets for consistency
    ss = StandardScaler()
    fea_scaled = ss.fit_transform(fea)
    
    # Select optimal features based on sigma type
    if sigma == 'promoter':
        optimfea = optim_fea_fromindep(fea_scaled, f'{model_dir}/promoter_pos.txt')
    else:
        optimfea = optim_fea_fromindep(fea_scaled, f'{model_dir}/sigma{sigma}_pos.txt')
    
    return optimfea
```
